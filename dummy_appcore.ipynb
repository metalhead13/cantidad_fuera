```python
import pandas as pd
import random
from faker import Faker

# Inicializar Faker
fake = Faker('es_ES')

# Lista de comerciales, clientes y países
BusinessDevelopmentSpecialist = ["Niolimar Sequera", "Alexander Gabel", "Víctor Díaz", "Angelica Garzón", "Andres Leon", "Claudía Navas"]
TechTalentPartner = ["Alexander Gabel", "Víctor Díaz", "Angelica Garzón", "Carolina Sabogal", "Alicia Perdomo", "Barbara Correa", "Cristian Carrillo"]
clientes = ["Banco Falabella", "Mi Banco", "Softtek", "BBVA", "Cliente A", "Cliente B", "Cliente C", "Cliente D", "Cliente E", "Cliente F", "Cliente G", "Cliente H", "Cliente I", "Cliente J", "Cliente K", "Cliente L", "Cliente M", "Cliente N", "Cliente O", "Cliente P", "Cliente Q", "Cliente R", "Cliente S", "Cliente T"]
paises = ["Colombia", "España"]

# Generar datos aleatorios para BASE CLIENTES
data_clientes = []
for _ in range(1000):
    marca_temporal = fake.date_time_between(start_date='-1y', end_date='now').strftime('%Y-%m-%d %H:%M:%S')
    pais = random.choice(paises)
    comercial = random.choice(BusinessDevelopmentSpecialist)
    cliente = random.choice(clientes)
    nombre_correo_gestor = fake.email()
    n_vacantes = random.randint(1, 5)
    perfil_puesto = fake.job()
    tiempo_servicio = random.choice(["1 Año", "> 1 Año", "Indefinido", "6 meses", "Más de 1 año"])
    modalidad_trabajo = random.choice(["Hibrido", "100 % Remoto", "Presencial"])
    formacion_academica = random.choice(["Ingeniería de Sistemas", "Tecnología", "Pregrado Ingeniería de Sistemas"])
    conocimientos_tecnicos = fake.text(max_nb_chars=200)
    años_experiencia = random.choice(["6 meses - 1 año", "> a 3 años", "5 Años"])
    asignacion_equipo = random.choice(["Cliente", "Samtel"])
    especificaciones_maquina = random.choice(["N/A", "Por determinar", "Por definir"])
    tarifa = random.randint(1000000, 10000000)
    salario = f"$ {random.randint(1000000, 5000000)} + ${random.randint(100000, 500000)}"
    observaciones = fake.text(max_nb_chars=100)
    usuario = random.choice(TechTalentPartner)
    estado = random.choice(["ATENDIDA", "CERRADA", "CONTRATADO"])
    observacion_proceso = fake.text(max_nb_chars=50)
    n_candidatos_enviados = random.randint(1, 10)
    fecha_envio = fake.date_time_between(start_date='-1y', end_date='now').strftime('%Y-%m-%d %H:%M:%S')
    candidatos = [fake.name() for _ in range(6)]

    data_clientes.append([
        marca_temporal, pais, comercial, cliente, nombre_correo_gestor, n_vacantes, perfil_puesto, tiempo_servicio,
        modalidad_trabajo, formacion_academica, conocimientos_tecnicos, años_experiencia, asignacion_equipo,
        especificaciones_maquina, tarifa, salario, observaciones, usuario, estado, observacion_proceso,
        n_candidatos_enviados, fecha_envio, *candidatos
    ])

# Crear DataFrame para BASE CLIENTES
df_clientes = pd.DataFrame(data_clientes, columns=[
    "Marca temporal", "Pais", "Comercial encargado", "Cliente", "Nombre-Correo Gestor Cliente", "N° Vacantes",
    "Perfil del Puesto", "Tiempo del Servicio", "Modalidad de Trabajo", "Formación Académica Mínima Requerida",
    "Conocimientos Técnicos Específicos", "Años y Tipo de Experiencia", "Asignación Equipo",
    "Especificaciones Máquina Portátil", "Tarifa", "Salario", "Observaciones", "USUARIO", "Estado",
    "Observación Proceso", "Número de Candidatos Enviados", "Fecha de Envío al Solicitante",
    "Candidato 1", "Candidato 2", "Candidato 3", "Candidato 4", "Candidato 5", "Candidato 6"
])

# Guardar en Excel
df_clientes.to_excel("Datos_Clientes.xlsx", index=False)

# Guardar en TXT
df_clientes.to_csv("Datos_Clientes.txt", sep="\t", index=False)

print("Archivos generados: Datos_Clientes.xlsx y Datos_Clientes.txt")
```
